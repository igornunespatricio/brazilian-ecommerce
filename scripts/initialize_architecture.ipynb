```python
%sql
-- create brazilian ecommerce catalog
create catalog if not exists brazilian_ecommerce
comment 'This is a Brazilian ecommerce public dataset of orders made at Olist Store. The dataset has information of 100k orders from 2016 to 2018 made at multiple marketplaces in Brazil. Its features allows viewing an order from multiple dimensions: from order status, price, payment and freight performance to customer location, product attributes and finally reviews written by customers.'
```

```python
%sql
create schema if not exists brazilian_ecommerce.prod
comment 'Production schema'
```

```python
%sql
create volume if not exists brazilian_ecommerce.prod.raw_data
comment 'Volume storing the raw dataset files'
```
